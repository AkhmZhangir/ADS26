```python
import sys


class BST:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

    def insert(self, num):
        current = self

        while True:
            if num < current.val:
                if current.left is None:
                    current.left = BST(num)
                    return

                current = current.left

            elif num > current.val:
                if current.right is None:
                    current.right = BST(num)
                    return

                current = current.right

            else:
                return

    def search(self, num):
        current = self

        while current is not None:
            if current.val == num:
                return current

            if num < current.val:
                current = current.left
            else:
                current = current.right

        return None

    def sub(self):
        count = 0
        stack = [self]

        while stack:
            current = stack.pop()
            count += 1

            if current.left is not None:
                stack.append(current.left)

            if current.right is not None:
                stack.append(current.right)

        return count
    def subb(self, pos):
        current = self
        res = []

        # Ищем узел pos
        while current.val != pos:
            if current.val < pos:
                current = current.right
            else:
                current = current.left

        # Обход поддерева в порядке:
        # node -> left -> right
        stack = [current]

        while stack:
            current = stack.pop()
            res.append(current.val)

            # Важно: right добавляем ПЕРВЫМ
            if current.right is not None:
                stack.append(current.right)

            # left добавляем ВТОРЫМ,
            # значит он выйдет из стека раньше
            if current.left is not None:
                stack.append(current.left)

        return res

n = int(sys.stdin.readline())
nums = list(map(int, sys.stdin.readline().split()))
target = int(sys.stdin.readline())

root = BST(nums[0])

for x in nums[1:]:
    root.insert(x)

result=root.subb(target)

print(*result)
```

```python
import sys

def main():
    data = sys.stdin.read().split()
    n = int(data[0])
    a = list(map(int, data[1:1 + n]))

    # BST на массивах (без рекурсии, чтобы не словить RecursionError)
    left = [-1] * n
    right = [-1] * n
    val = [0] * n
    val[0] = a[0]
    size = 1

    for x in a[1:]:
        cur = 0
        while True:
            if x < val[cur]:
                if left[cur] == -1:
                    val[size] = x
                    left[cur] = size
                    size += 1
                    break
                cur = left[cur]
            elif x > val[cur]:
                if right[cur] == -1:
                    val[size] = x
                    right[cur] = size
                    size += 1
                    break
                cur = right[cur]
            else:
                break  # дубликат пропускаем

    count = 0
    for i in range(size):
        if left[i] != -1 and right[i] != -1:
            count += 1
    print(count)

main()
```

```python
import sys

def main():
    data = sys.stdin.read().split()
    n = int(data[0])
    raw = list(map(int, data[1:1 + n]))

    seen = set()
    a = []
    for x in raw:
        if x not in seen:
            seen.add(x)
            a.append(x)
    n = len(a)

    order = sorted(range(n), key=lambda i: a[i])
    rank = [0] * n
    for r, t in enumerate(order):
        rank[t] = r

    prv = [r - 1 for r in range(n)]
    nxt = [r + 1 if r + 1 < n else -1 for r in range(n)]

    L = [-1] * n
    R = [-1] * n

    for t in range(n - 1, 0, -1):
        r = rank[t]
        p, s = prv[r], nxt[r]
        if p == -1:
            L[order[s]] = t
        elif s == -1:
            R[order[p]] = t
        elif order[p] > order[s]:
            R[order[p]] = t
        else:
            L[order[s]] = t
        if p != -1:
            nxt[p] = s
        if s != -1:
            prv[s] = p

    h = [0] * n
    best = 0
    for t in range(n - 1, -1, -1):
        hl = h[L[t]] if L[t] != -1 else 0
        hr = h[R[t]] if R[t] != -1 else 0
        h[t] = 1 + max(hl, hr)
        if hl + hr + 1 > best:
            best = hl + hr + 1

    print(best)

main()
```

```python
import sys

class BST:
    def __init__(self, arr):
        seen = set()
        self.a = []
        for x in arr:
            if x not in seen:
                seen.add(x)
                self.a.append(x)
        self.n = len(self.a)

    def count_leaves(self):
        n, a = self.n, self.a
        if n == 0:
            return 0
        order = sorted(range(n), key=lambda i: a[i])
        rank = [0] * n
        for r, t in enumerate(order):
            rank[t] = r
        prv = [r - 1 for r in range(n)]
        nxt = [r + 1 if r + 1 < n else -1 for r in range(n)]
        is_parent = [False] * n

        for t in range(n - 1, 0, -1):
            r = rank[t]
            p, s = prv[r], nxt[r]
            if p == -1:
                is_parent[order[s]] = True
            elif s == -1:
                is_parent[order[p]] = True
            elif order[p] > order[s]:
                is_parent[order[p]] = True
            else:
                is_parent[order[s]] = True
            if p != -1:
                nxt[p] = s
            if s != -1:
                prv[s] = p

        return n - sum(is_parent)

def main():
    data = sys.stdin.read().split()
    n = int(data[0])
    tree = BST(list(map(int, data[1:1 + n])))
    print(tree.count_leaves())

main()
```
