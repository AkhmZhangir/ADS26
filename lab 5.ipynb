```python
import heapq,sys
a,b = map(int,sys.stdin.readline().split())
c=[0]
s=0
heapq.heapify(c)
for i in range(a):
    q=sys.stdin.readline().split()
    if q[0]=="print":
        print(s)
        continue
    elif q[0]=="insert":
        if len(c)<b:
            heapq.heappush(c,int(q[1]))
            s+=int(q[1])
        else:
            g=heapq.heappop(c)
            s-=g
            if g<int(q[1]):
                heapq.heappush(c,int(q[1]))
                s+=int(q[1])
            else:
                heapq.heappush(c,g)
                s+=int(q[1])
        continue
```
