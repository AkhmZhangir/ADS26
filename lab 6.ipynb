```python
import sys
from collections import deque

a, b = map(int, sys.stdin.readline().split())
if a > 0:
    c = deque(map(int, sys.stdin.readline().split()))
else:
    exit()
if b > 0:
    d = deque(map(int, sys.stdin.readline().split()))
else:
    exit()
c = deque(sorted(c))
d = deque(sorted(d))
i = 0
j = 0
while j < min(a, b):
    if c[i] == d[i]:
        print(c[i], end=' ')
        c.popleft()
        d.popleft()
    else:
        if c[i] > d[i]:
            d.popleft()
        else:
            c.popleft()
    if min(len(c), len(d)) == 0:
        break
```

```python
a=int(input())
gpa=list()
names=list()
students=list()
for i in range(a):
    s=input().split()
    last_name=s[0]
    first_name=s[1]
    names.append(last_name)
    names.append(first_name)
    subjects=int(s[2])
    d=s[3:]
    for j in range(0,len(d),2):
        if d[j]=='A+':
            d[j]=float(4)
        elif d[j]=='A':
            d[j]=float(3.75)
        elif d[j]=='B+':
            d[j]=float(3.5)
        elif d[j]=='B':
            d[j]=float(3.0)
        elif d[j]=='C+':
            d[j]=float(2.5)
        elif d[j]=='C':
            d[j]=float(2.0)
        elif d[j]=='D+':
            d[j]=float(1.5)
        elif d[j]=='D':
            d[j]=float(1.0)
        elif d[j]=='F':
            d[j]=float(0.0)
    su = 0
    cre = 0

    for j in range(0, len(d), 2):
        grade = float(d[j])
        credit = int(d[j + 1])

        su += grade * credit
        cre += credit

    students.append([su / cre, last_name, first_name])
students.sort()
for val, last,first in students:
    print(f"{last} {first} {val:.3f}")
```

```python
import heapq
def heap_sort(arr):
    heapq.heapify(arr)
    ans= list()
    while len(arr)>0:
        ans.append(heapq.heappop(arr))
    return ans
def quick_sort(arr, left, right):
    if left >= right:
        return

    pivot = arr[(left + right) // 2]
    lt = left
    i = left
    gt = right

    while i <= gt:
        if arr[i] < pivot:
            arr[i], arr[lt] = arr[lt], arr[i]
            lt += 1
            i += 1

        elif arr[i] > pivot:
            arr[i], arr[gt] = arr[gt], arr[i]
            gt -= 1

        else:
            i += 1

    quick_sort(arr, left, lt - 1)
    quick_sort(arr, gt + 1, right)


arr = list(input())
quick_sort(arr, 0, len(arr) - 1)
print("".join(arr))
```
